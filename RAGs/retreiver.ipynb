```python
from langchain_community.retrievers import WikipediaRetriever
import wikipedia
```

```python
# wikipedia.set_user_agent("MyResearchProject/1.0 (contact@example.com)")

retreiver = WikipediaRetriever(top_k_results=2, lang="en")
```

```python
query = "the political history of Pakistan and India"

docs = retreiver.invoke(query)
```
