```python
from langchain_core.documents import Document
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from dotenv import load_dotenv
import os

HF_TOKEN = os.environ["HUGGINGFACEHUB_API_TOKEN"]

load_dotenv()

doc1 = Document(
    page_content = "Shahid Khan Afridi, a really cool player of the PAK. Also known as BOOM BOOM. His main trait was his aggressive batting style. he also excels at bowling a gowgly",
    metadata = {"team" : "Peshawar Zalmi"}
)

doc2 = Document(
    page_content = "Shoaib Akhtar, a really cool player of the PAK. Also known as Rawalpindi Express. His main trait was his aggressive bowling style. he was short temperred",
    metadata = {"team" : "Quetta Gladiators"}
)

doc3 = Document(
    page_content = "Waseem Akram, a really cool player of the PAK. Also known as BOOM BOOM. His main trait was his aggressive batting style. he also excels at bowling a gowgly",
    metadata = {"team" : "Karachi Kings"}
)

doc4 = Document(
    page_content = "Muhammad Amir, a really cool player of the PAK. Also known as BOOM BOOM. His main trait was his aggressive batting style. he also excels at bowling a gowgly",
    metadata = {"team" : "Lahore Qalandars"}
)
```

```python
docs = [doc1, doc2, doc3, doc4]
```

```python
vector_store = Chroma(
    embedding_function = HuggingFaceEmbeddings(model_name="BAAI/bge-base-en-v1.5",  model_kwargs={"token": HF_TOKEN}),
    persist_directory="my_chroma_db",
    collection_name="sample"
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['documents'])
```

```python
vector_store.similarity_search(
    query="Who among them is also known as rawalpindi express",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="Who among them is also known as rawalpindi express",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="",
    filter={
        "team": "Karachi Kings"
    }
)
```
